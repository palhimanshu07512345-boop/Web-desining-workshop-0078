```python
from IPython.core.display import display, HTML

html_code = """<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">

<title>Friendship Proposal for Saloni ❤️</title>

<style>
* {
    box-sizing: border-box;
}

body {
    margin: 0;
    height: 100vh;
    overflow: hidden;
    background: #ffe4ef;
    display: flex;
    justify-content: center;
    align-items: center;
    font-family: Arial, sans-serif;
}

/* Floating Hearts */

.heart {
    position: absolute;
    bottom: -30px;
    animation: floatUp linear infinite;
    pointer-events: none;
}

@keyframes floatUp {
    0% {
        transform: translateY(0);
        opacity: 1;
    }
    100% {
        transform: translateY(-110vh);
        opacity: 0;
    }
}

/* Main Card */

.card {
    width: 90%;
    max-width: 420px;
    min-height: 530px;
    background: white;
    border-radius: 25px;
    box-shadow: 0 10px 30px rgba(0,0,0,0.15);
    text-align: center;
    padding: 25px;
    position: relative;
    z-index: 2;
}

.big-heart {
    font-size: 45px;
    animation: heartbeat 0.8s infinite alternate;
}

@keyframes heartbeat {
    from { transform: scale(1); }
    to { transform: scale(1.2); }
}

h1 {
    color: #e75480;
    font-size: 25px;
}

#story {
    color: #555;
    font-size: 16px;
    line-height: 1.6;
    min-height: 100px;
    margin-top: 30px;
}

/* Buttons */

button {
    border: none;
    border-radius: 10px;
    padding: 12px 22px;
    font-size: 15px;
    font-weight: bold;
    cursor: pointer;
    margin: 10px;
    transition: 0.2s;
}

button:hover {
    transform: scale(1.05);
}

#nextBtn {
    background: #ff69a4;
    color: white;
}

#yesBtn {
    background: #ff4f91;
    color: white;
    font-size: 17px;
}

#noBtn {
    background: #eee;
    color: #444;
    position: absolute;
}

#shareBtn {
    background: #25D366;
    color: white;
}

.hidden {
    display: none;
}
</style>
</head>

<body>

<div class="card">

    <div class="big-heart">💗</div>

    <h1 id="title">Hey Saloni! ❤️</h1>

    <p id="story">
        Ek special surprise aapkee liye madam ji...
    </p>

    <button id="nextBtn" onclick="nextPage()">
        Start Surprise 💞
    </button>

    <button id="yesBtn" class="hidden" onclick="yesClick()">
        YES ❤️
    </button>

    <button id="noBtn" class="hidden" onclick="moveNo()">
        NO 🙈
    </button>

    <button id="shareBtn" class="hidden"
        onclick="shareWhatsApp()">
        Share on WhatsApp 💚
    </button>

</div>

<script>

// Friendship Chapters

const chapters = [

    ["Our First Meeting 🥰",
    "Pata nahi tha ek din tum itne special dost ban jaoge!"],

    ["Crazy Memories 😂",
    "Hamari random baatein aur bina wajah hasi!"],

    ["Best Support 🤝",
    "Har situation mein ek dusre ka saath."],

    ["Secret Keeper 🤫",
    "Tumhare saath har baat share kar sakta hoon."],

    ["8 Years Together 💗",
    "8 saal ki dosti aur hazaaron yaadein!"],

    ["Forever Friendship ❤️",
    "Abhi toh bahut saari memories banana baaki hain!"],

    ["My Secret Keeper 🥰",
    "Kuch bate sirf tumhare saath hi share kar sakta hoon!"],

    ["Our First Meetup ❤️",
    "Lucknow mein vo hamari mulakat aaj bhi yaad hai aur hamesha yaad rahegi!"],

    ["No Distance Can Change Us 🙈",
    "Dooriya badh sakti hain par ham logon ki dosti kabhi nahi!"],

    ["Friends, Not Lovers 🙈",
    "Pyar se badhkar bhi ek rishta hota hai aur vo hamari unbreakable dosti hai!"],

    ["Our Future Plans 🥰",
    "Abhi to ek trip maarni hai hum logon ko aur vo bhi jald hi!"]

];

let page = 0;

// Elements

const title = document.getElementById("title");
const story = document.getElementById("story");

const nextBtn = document.getElementById("nextBtn");
const yesBtn = document.getElementById("yesBtn");
const noBtn = document.getElementById("noBtn");
const shareBtn = document.getElementById("shareBtn");

// Next Page Function

function nextPage() {

    if (page < chapters.length) {

        title.innerText = chapters[page][0];
        story.innerText = chapters[page][1];

        page++;

        nextBtn.innerText = "Next Surprise 💌";

    } else {

        title.innerText = "One Question! 🥹";

        story.innerText =
        "Kya hamari friendship lifetime tak aise hi rahegi?";

        nextBtn.classList.add("hidden");

        yesBtn.classList.remove("hidden");
        noBtn.classList.remove("hidden");

        moveNo();
    }
}

// YES Button

function yesClick() {

    title.innerText = "YAYYYYY! 🥹❤️";

    story.innerText =
    "Best Friendship Forever! 💗\\nThank you for being my best friend!";

    yesBtn.classList.add("hidden");
    noBtn.classList.add("hidden");

    shareBtn.classList.remove("hidden");
}

// NO Button Animation

function moveNo() {

    if (noBtn.classList.contains("hidden")) return;

    const card = document.querySelector(".card");

    const maxX = card.clientWidth - noBtn.offsetWidth - 20;
    const maxY = card.clientHeight - noBtn.offsetHeight - 20;

    const x = Math.random() * Math.max(0, maxX);
    const y = Math.random() * Math.max(0, maxY);

    noBtn.style.left = x + "px";
    noBtn.style.top = y + "px";
}

// WhatsApp Sharing

function shareWhatsApp() {

    const message =
    "Hey Bestie ❤️\\n" +
    "Mere paas tumhare liye ek friendship surprise hai! 🥹";

    const link =
    "https://wa.me/?text=" +
    encodeURIComponent(message);

    window.open(link, "_blank");

}

// Floating Hearts Animation

function createHeart() {

    const heart = document.createElement("div");

    heart.className = "heart";
    heart.innerText = "❤️";

    heart.style.left = Math.random() * 100 + "vw";

    heart.style.fontSize =
    (12 + Math.random() * 15) + "px";

    heart.style.animationDuration =
    (4 + Math.random() * 5) + "s";

    document.body.appendChild(heart);

    setTimeout(() => {
        heart.remove();
    }, 9000);
}

setInterval(createHeart, 400);

// Initial Hearts

for (let i = 0; i < 15; i++) {
    createHeart();
}

</script>

</body>
</html>"""

display(HTML(html_code))
```
